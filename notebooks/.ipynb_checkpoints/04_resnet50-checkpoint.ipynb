# 04. ResNet50 con Transfer Learning

En este notebook se entrena una arquitectura ResNet50 preentrenada en ImageNet para clasificar imágenes de TC pulmonar en tres clases: `benign`, `malignant` y `normal`.

El objetivo es evaluar el rendimiento de una red residual profunda mediante transfer learning, utilizando las mismas particiones de entrenamiento, validación y prueba que en el resto de modelos.

El entrenamiento se realiza en dos fases:

1. Entrenamiento de una cabeza clasificadora con la base ResNet50 congelada.
2. Fine-tuning parcial de las últimas capas de la red.

Los resultados obtenidos se guardarán para su posterior comparación con MobileNetV2 y EfficientNetB0.

## 0. Configuración inicial

Se importan las librerías necesarias, se fija la semilla de reproducibilidad y se definen los parámetros principales del experimento: tamaño de entrada, batch size, número de épocas y clases del problema.cientNetB0.

```python
import os
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, optimizers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet50_preprocess_input

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_recall_fscore_support
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

IMG_SIZE = (224, 224)
BATCH_SIZE = 16

EPOCHS_HEAD = 15
EPOCHS_FINE = 20

CLASSES = ["benign", "malignant", "normal"]

print("TensorFlow:", tf.__version__)
print("GPU disponible:", tf.config.list_physical_devices("GPU"))
```

## 1. Localización de archivos procesados

Se localizan los archivos generados durante el análisis exploratorio del dataset. Este notebook reutiliza las particiones `train.csv`, `val.csv` y `test.csv`.

No se realiza una nueva división del dataset, ya que todos los modelos deben evaluarse sobre las mismas particiones para que la comparación sea válida.t.

```python
possible_processed_paths = [
    Path("/kaggle/working/processed"),
    Path("/content/processed"),
    Path("processed"),
    Path("../processed"),
    Path("../data/processed"),
    Path("data/processed")
]

processed_path = None

for candidate_path in possible_processed_paths:
    if (
        (candidate_path / "train.csv").exists()
        and (candidate_path / "val.csv").exists()
        and (candidate_path / "test.csv").exists()
    ):
        processed_path = candidate_path
        break

if processed_path is None:
    raise FileNotFoundError(
        "No se encontraron train.csv, val.csv y test.csv. "
        "Ejecuta primero el notebook 01_eda_dataset.ipynb o copia la carpeta processed."
    )

if Path("/kaggle/working").exists():
    output_path = Path("/kaggle/working/resnet50_outputs")
elif Path("/content").exists():
    output_path = Path("/content/resnet50_outputs")
else:
    output_path = Path("../outputs/resnet50_outputs")

models_path = output_path / "models"
figures_path = output_path / "figures"
metrics_path = output_path / "metrics"

for path in [output_path, models_path, figures_path, metrics_path]:
    path.mkdir(parents=True, exist_ok=True)

print("Carpeta processed:", processed_path)
print("Carpeta de salida:", output_path)
```

## 2. Carga de particiones

Se cargan los conjuntos de entrenamiento, validación y prueba. Estas particiones proceden del EDA y son comunes para todos los modelos evaluados en el proyecto.2.

```python
train_df = pd.read_csv(processed_path / "train.csv")
val_df = pd.read_csv(processed_path / "val.csv")
test_df = pd.read_csv(processed_path / "test.csv")

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nDistribución train:")
print(train_df["class"].value_counts())

print("\nDistribución validation:")
print(val_df["class"].value_counts())

print("\nDistribución test:")
print(test_df["class"].value_counts())

train_df.head()
```

```python
for split_name, split_df in {
    "train": train_df,
    "validation": val_df,
    "test": test_df
}.items():
    missing_paths = split_df[~split_df["path"].apply(lambda x: Path(x).exists())]
    print(f"{split_name} - rutas inexistentes: {len(missing_paths)}")
```

## 3. Generadores de imágenes

Se crean los generadores de imágenes específicos para ResNet50. Las imágenes se redimensionan a 224x224 píxeles y se procesan mediante la función `preprocess_input` propia de ResNet50.

En el conjunto de entrenamiento se aplica aumentación moderada para reducir el sobreajuste. En validación y test no se aplica aumentación, únicamente el preprocesamiento necesario.t50.

```python
train_datagen = ImageDataGenerator(
    preprocessing_function=resnet50_preprocess_input,
    rotation_range=10,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.10,
    horizontal_flip=True,
    fill_mode="nearest"
)

val_test_datagen = ImageDataGenerator(
    preprocessing_function=resnet50_preprocess_input
)

train_gen = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col="path",
    y_col="class",
    classes=CLASSES,
    target_size=IMG_SIZE,
    color_mode="rgb",
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED
)

val_gen = val_test_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col="path",
    y_col="class",
    classes=CLASSES,
    target_size=IMG_SIZE,
    color_mode="rgb",
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_gen = val_test_datagen.flow_from_dataframe(
    dataframe=test_df,
    x_col="path",
    y_col="class",
    classes=CLASSES,
    target_size=IMG_SIZE,
    color_mode="rgb",
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Índices de clase:", train_gen.class_indices)
```

## 4. Cálculo de pesos de clase

Debido al desbalanceo entre clases, especialmente en la clase `benign`, se calculan pesos de clase usando únicamente el conjunto de entrenamiento.

Estos pesos permiten que los errores cometidos en clases minoritarias tengan mayor impacto durante el proceso de aprendizaje.o.

```python
class_names = np.array(CLASSES)

weights = compute_class_weight(
    class_weight="balanced",
    classes=class_names,
    y=train_df["class"]
)

class_weight_by_name = dict(zip(class_names, weights))

class_weight = {
    train_gen.class_indices[class_name]: weight
    for class_name, weight in class_weight_by_name.items()
}

print("Pesos por clase:")
print(class_weight_by_name)

print("\nPesos por índice Keras:")
print(class_weight)
```

## 5. Definición del modelo ResNet50

Se utiliza ResNet50 como base convolucional preentrenada en ImageNet. La capa superior original se elimina y se añade una cabeza clasificadora adaptada a las tres clases del problema.

En esta primera fase, la base ResNet50 permanece congelada y solo se entrena la nueva cabeza clasificadora.ada.

```python
def build_resnet50_model(input_shape=(224, 224, 3), num_classes=3):
    base_model = ResNet50(
        include_top=False,
        weights="imagenet",
        input_shape=input_shape
    )

    base_model.trainable = False

    inputs = layers.Input(shape=input_shape)

    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.4)(x)

    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = models.Model(inputs, outputs)

    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model, base_model


resnet50_model, resnet50_base = build_resnet50_model(
    input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3),
    num_classes=len(CLASSES)
)

resnet50_model.summary()
```

## 6. Entrenamiento de la cabeza clasificadora

En esta fase se entrena únicamente la cabeza clasificadora añadida sobre ResNet50. La base convolucional permanece congelada para aprovechar las características generales aprendidas durante el preentrenamiento.

Se utilizan callbacks para guardar el mejor modelo, reducir el learning rate cuando la validación se estanca y detener el entrenamiento si no se observa mejora.a.

```python
resnet50_head_checkpoint_path = models_path / "resnet50_head_best.keras"

head_callbacks = [
    callbacks.ModelCheckpoint(
        filepath=resnet50_head_checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-7,
        verbose=1
    ),
    callbacks.EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True,
        verbose=1
    )
]

history_head = resnet50_model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=EPOCHS_HEAD,
    callbacks=head_callbacks,
    class_weight=class_weight,
    verbose=1
)
```

## 7. Fine-tuning del modelo

Tras entrenar la cabeza clasificadora, se desbloquean las últimas capas de ResNet50 para adaptar parcialmente el modelo al dominio de las imágenes de TC pulmonar.

El fine-tuning se realiza con un learning rate bajo para evitar modificar de forma brusca los pesos preentrenados.g.

```python
resnet50_base.trainable = True

for layer in resnet50_base.layers[:-30]:
    layer.trainable = False

for layer in resnet50_base.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

resnet50_model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

trainable_layers = sum([1 for layer in resnet50_base.layers if layer.trainable])
print("Capas entrenables en la base ResNet50:", trainable_layers)
```

```python
resnet50_finetune_checkpoint_path = models_path / "resnet50_finetuned_best.keras"

finetune_callbacks = [
    callbacks.ModelCheckpoint(
        filepath=resnet50_finetune_checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-8,
        verbose=1
    ),
    callbacks.EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True,
        verbose=1
    )
]

history_fine = resnet50_model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=EPOCHS_FINE,
    callbacks=finetune_callbacks,
    class_weight=class_weight,
    verbose=1
)
```

## 8. Curvas de entrenamiento

Se guardan y visualizan las curvas de pérdida y accuracy durante las fases de entrenamiento de la cabeza clasificadora y fine-tuning.

Estas curvas permiten comprobar si el modelo aprende de forma estable y si aparece sobreajuste.o.

```python
history_head_df = pd.DataFrame(history_head.history)
history_fine_df = pd.DataFrame(history_fine.history)

history_head_df["phase"] = "head"
history_fine_df["phase"] = "fine_tuning"

history_resnet50_df = pd.concat(
    [history_head_df, history_fine_df],
    ignore_index=True
)

history_resnet50_df.to_csv(metrics_path / "resnet50_history.csv", index=False)

display(history_resnet50_df.head())
```

```python
plt.figure(figsize=(7, 4))
plt.plot(history_resnet50_df["loss"], label="train_loss")
plt.plot(history_resnet50_df["val_loss"], label="val_loss")
plt.title("ResNet50 - Pérdida")
plt.xlabel("Época")
plt.ylabel("Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig(figures_path / "resnet50_loss_curve.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.plot(history_resnet50_df["accuracy"], label="train_accuracy")
plt.plot(history_resnet50_df["val_accuracy"], label="val_accuracy")
plt.title("ResNet50 - Accuracy")
plt.xlabel("Época")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig(figures_path / "resnet50_accuracy_curve.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 9. Evaluación en el conjunto de test

El conjunto de test se utiliza únicamente al final del entrenamiento. Se evalúa el mejor modelo guardado durante el fine-tuning para obtener una estimación final de rendimiento.e.

```python
best_resnet50_model = tf.keras.models.load_model(resnet50_finetune_checkpoint_path)

test_loss, test_accuracy = best_resnet50_model.evaluate(test_gen, verbose=1)

print("Test loss ResNet50:", test_loss)
print("Test accuracy ResNet50:", test_accuracy)
```

## 10. Métricas por clase

Se calculan precision, recall y F1-score para cada clase. Estas métricas son especialmente importantes porque el dataset está desbalanceado y la accuracy por sí sola puede resultar insuficiente.

Se prestará especial atención al comportamiento de la clase `malignant`, ya que los falsos negativos en esta clase son los errores más críticos.

```python
test_gen.reset()

y_prob = best_resnet50_model.predict(test_gen, verbose=1)
y_pred = np.argmax(y_prob, axis=1)
y_true = test_gen.classes

idx_to_class = {
    v: k for k, v in test_gen.class_indices.items()
}

target_names = [
    idx_to_class[i]
    for i in range(len(idx_to_class))
]

report_dict = classification_report(
    y_true,
    y_pred,
    target_names=target_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report_dict).transpose()

display(report_df)

report_df.to_csv(metrics_path / "resnet50_classification_report.csv")
```

## 11. Métricas resumen

Se calculan métricas globales del modelo, incluyendo accuracy, macro precision, macro recall, macro F1-score y weighted F1-score.

El macro F1-score es una métrica relevante en este problema porque otorga el mismo peso a todas las clases, independientemente del número de muestras de cada una.

```python
macro_precision, macro_recall, macro_f1, _ = precision_recall_fscore_support(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

weighted_precision, weighted_recall, weighted_f1, _ = precision_recall_fscore_support(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

summary_metrics = pd.DataFrame({
    "model": ["ResNet50"],
    "test_loss": [test_loss],
    "test_accuracy": [accuracy_score(y_true, y_pred)],
    "macro_precision": [macro_precision],
    "macro_recall": [macro_recall],
    "macro_f1": [macro_f1],
    "weighted_precision": [weighted_precision],
    "weighted_recall": [weighted_recall],
    "weighted_f1": [weighted_f1]
})

display(summary_metrics)

summary_metrics.to_csv(metrics_path / "resnet50_summary_metrics.csv", index=False)
```

## 12. Matriz de confusión

Se genera la matriz de confusión para analizar qué clases se confunden entre sí.

En este problema, los errores más relevantes son los casos `malignant` clasificados como `normal`, ya que representan falsos negativos clínicamente críticos.s.

```python
cm = confusion_matrix(y_true, y_pred)

cm_df = pd.DataFrame(
    cm,
    index=target_names,
    columns=target_names
)

display(cm_df)

cm_df.to_csv(metrics_path / "resnet50_confusion_matrix.csv")
```

```python
plt.figure(figsize=(6, 5))
plt.imshow(cm)

plt.title("ResNet50 - Matriz de confusión")
plt.xlabel("Predicción")
plt.ylabel("Etiqueta real")

plt.xticks(
    ticks=np.arange(len(target_names)),
    labels=target_names,
    rotation=45
)

plt.yticks(
    ticks=np.arange(len(target_names)),
    labels=target_names
)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, str(cm[i, j]), ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(figures_path / "resnet50_confusion_matrix.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 13. Distribución de predicciones

Se analiza cuántas imágenes han sido asignadas a cada clase por el modelo. Esta comprobación permite verificar que ResNet50 predice las tres categorías y no colapsa hacia una única clase.s.

```python
predicted_class_counts = pd.Series(
    [idx_to_class[i] for i in y_pred]
).value_counts()

print("Distribución de predicciones ResNet50:")
print(predicted_class_counts)
```

```python
probabilities_summary = pd.DataFrame(
    y_prob,
    columns=[idx_to_class[i] for i in range(len(idx_to_class))]
).describe()

display(probabilities_summary)

probabilities_summary.to_csv(metrics_path / "resnet50_probabilities_summary.csv")
```

## 14. Guardado de predicciones

Se guarda un archivo con la clase real, la clase predicha, la confianza del modelo y las probabilidades asociadas a cada clase para todas las imágenes del conjunto de test.

Este archivo será utilizado posteriormente para comparar modelos y analizar errores de clasificación.t.

```python
predictions_df = test_df.copy().reset_index(drop=True)

predictions_df["true_class"] = [
    idx_to_class[i]
    for i in y_true
]

predictions_df["pred_class"] = [
    idx_to_class[i]
    for i in y_pred
]

predictions_df["confidence"] = np.max(y_prob, axis=1)

predictions_df["correct"] = (
    predictions_df["true_class"] == predictions_df["pred_class"]
)

for class_name, class_idx in test_gen.class_indices.items():
    predictions_df[f"prob_{class_name}"] = y_prob[:, class_idx]

predictions_df.to_csv(metrics_path / "resnet50_test_predictions.csv", index=False)

display(predictions_df.head())
```

## 15. Análisis básico de errores

Se identifican las imágenes mal clasificadas por ResNet50. En este notebook solo se realiza una inspección inicial, ya que el análisis comparativo detallado se llevará a cabo en el notebook final de comparación de modelos.s.

```python
errors_df = predictions_df[~predictions_df["correct"]].copy()

print("Número de errores en test:", len(errors_df))

display(
    errors_df[
        ["filename", "true_class", "pred_class", "confidence", "path"]
    ].head(20)
)
```

```python
if len(errors_df) > 0:
    sample_errors = errors_df.sample(
        n=min(9, len(errors_df)),
        random_state=SEED
    )

    plt.figure(figsize=(12, 8))

    for i, (_, row) in enumerate(sample_errors.iterrows()):
        img = tf.keras.utils.load_img(
            row["path"],
            target_size=IMG_SIZE
        )

        plt.subplot(3, 3, i + 1)
        plt.imshow(img)
        plt.title(
            f"Real: {row['true_class']}\n"
            f"Pred: {row['pred_class']} ({row['confidence']:.2f})"
        )
        plt.axis("off")

    plt.tight_layout()
    plt.savefig(figures_path / "resnet50_error_examples.png", dpi=150, bbox_inches="tight")
    plt.show()

else:
    print("No se encontraron errores de clasificación en test.")
```
